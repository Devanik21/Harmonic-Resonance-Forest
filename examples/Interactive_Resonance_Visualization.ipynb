# Interactive Resonance Visualization

This notebook provides an interactive demonstration of the Harmonic Resonance Forest (HRF) explainability tools. We will visualize class-wise resonance contributions, resonance intensity heatmaps, and decision boundaries.

```python
import sys
import os
sys.path.append(os.path.abspath('..'))

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split

from visualizations.resonance_plots import plot_class_resonance, plot_resonance_heatmap
from visualizations.decision_boundary import plot_decision_boundary

# Ensure plots are displayed inline
%matplotlib inline
```

## 1. Setup Mock HRF Model and Data
For this demonstration, we'll create a synthetic dataset and use a mock HRF model to simulate predictions and probabilities, similar to how HRF aggregates resonance energy.

```python
# Generate a synthetic classification dataset
X, y = make_classification(n_samples=200, n_features=2, n_informative=2, n_redundant=0, 
                           n_classes=3, n_clusters_per_class=1, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

class MockHRFModel:
    def __init__(self):
        self.classes_ = np.array([0, 1, 2])
    
    def fit(self, X, y):
        from sklearn.neighbors import KNeighborsClassifier
        self.model = KNeighborsClassifier(n_neighbors=5)
        self.model.fit(X, y)
        return self
        
    def predict_proba(self, X):
        # Simulate resonance probabilities using a simple KNN proxy for the demo
        return self.model.predict_proba(X)
        
    def predict(self, X):
        return self.model.predict(X)

model = MockHRFModel()
model.fit(X_train, y_train)
print("Mock HRF Model trained on synthetic data.")
```

## 2. Class-wise Resonance Contribution
Visualizing the contribution of each class's resonance for a specific prediction.

```python
sample_index = 0
sample = X_test[sample_index:sample_index+1]

fig = plot_class_resonance(model, sample, figsize=(8, 5))
plt.show()
```

## 3. Resonance Intensity Heatmap
Visualize resonance intensity patterns across a batch of samples and predicted classes.

```python
fig = plot_resonance_heatmap(model, X_test[:30], figsize=(10, 6))
plt.show()
```

## 4. Decision Boundary Visualization
Visualize how the model separates different regions based on accumulated resonance.

```python
fig, pca_transform = plot_decision_boundary(model, X_train, y_train, figsize=(10, 8), resolution=100)
plt.show()
```
